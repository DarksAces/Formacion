# 1. ¿Qué es un programa informático?

Un **programa informático** es un conjunto ordenado de instrucciones escritas en un lenguaje de programación. Estas instrucciones indican al ordenador qué datos debe recibir, qué operaciones debe realizar y qué resultado debe producir.

De forma simplificada, muchos programas siguen este esquema:

1. **Entrada:** datos que recibe el programa.
2. **Procesamiento:** operaciones realizadas con esos datos.
3. **Salida:** resultado generado.

## Ejemplo

Una cámara frigorífica registra una temperatura:

- **Entrada:** temperatura medida.
- **Procesamiento:** comparación con el intervalo permitido.
- **Salida:** aviso de temperatura correcta, baja o alta.

```python
# Ejemplo resuelto: control de temperatura

temperatura = 6

if temperatura < 2:
    resultado = "Temperatura demasiado baja"
elif temperatura > 8:
    resultado = "Temperatura demasiado alta"
else:
    resultado = "Temperatura correcta"

print(resultado)
```

- Muestre `"Temperatura demasiado alta"` si está por encima de 8.

```python
# Pruebas automáticas del ejercicio 2
# Ejecuta esta celda después de completar el ejercicio anterior.

def comprobar_temperatura(temperatura):
    if temperatura < -50 or temperatura > 60:
        return "Dato incorrecto"
    elif temperatura < 2:
        return "Temperatura demasiado baja"
    elif temperatura > 8:
        return "Temperatura demasiado alta"
    return "Temperatura correcta"

assert comprobar_temperatura(-60) == "Dato incorrecto"
assert comprobar_temperatura(0) == "Temperatura demasiado baja"
assert comprobar_temperatura(2) == "Temperatura correcta"
assert comprobar_temperatura(8) == "Temperatura correcta"
assert comprobar_temperatura(15) == "Temperatura demasiado alta"
assert comprobar_temperatura(70) == "Dato incorrecto"

print("Pruebas de referencia ejecutadas correctamente.")
```

# 2. Etapas del desarrollo de un programa

Crear un programa no consiste únicamente en escribir código. Habitualmente se siguen varias etapas:

1. **Definición del problema:** se concreta qué necesidad se quiere resolver.
2. **Análisis de requisitos:** se determina qué debe hacer el sistema y bajo qué condiciones.
3. **Diseño:** se decide cómo se organizarán los datos, los módulos, las interfaces y los algoritmos.
4. **Implementación:** se escribe el código.
5. **Pruebas:** se comprueba que el programa funciona y cumple los requisitos.
6. **Despliegue:** se instala o publica en el entorno donde se utilizará.
7. **Mantenimiento:** se corrigen errores y se incorporan mejoras.

# 4. Del requisito a la regla programable

Antes de utilizar Machine Learning, muchos problemas pueden resolverse mediante reglas definidas por una persona.

Para este primer prototipo se utilizarán estas reglas:

- Si el peso es igual o inferior a 0, el dato es incorrecto.
- Si el porcentaje de defectos está fuera del intervalo 0–100, el dato es incorrecto.
- Si los defectos superan el 30 %, la fruta se descarta.
- Si los defectos están entre el 11 % y el 30 %, se destina a procesamiento.
- Si los defectos son iguales o inferiores al 10 % y el peso es de al menos 150 g, es de primera categoría.
- En cualquier otro caso, es de segunda categoría.

```python
# Ejemplo resuelto

peso = 175
porcentaje_defectos = 8

if peso <= 0:
    categoria = "Dato incorrecto"
elif porcentaje_defectos < 0 or porcentaje_defectos > 100:
    categoria = "Dato incorrecto"
elif porcentaje_defectos > 30:
    categoria = "Descartada"
elif porcentaje_defectos > 10:
    categoria = "Procesamiento"
elif peso >= 150:
    categoria = "Primera categoría"
else:
    categoria = "Segunda categoría"

print(categoria)
```
