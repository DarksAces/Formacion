# 4. Funciones y separación de responsabilidades

Una función puede recibir datos mediante **parámetros** y devolver un resultado mediante `return`.

# 6. La función principal `main()`

Python no obliga a crear una función `main()`, pero suele ser una buena práctica.

Permite separar:

- Definiciones.
- Funciones auxiliares.
- Lógica principal.
- Ejecución.

## Estructura habitual

```python
def funcion_1():
    ...

def funcion_2():
    ...

def main():
    ...

if __name__ == "__main__":
    main()
```

La condición final evita que el programa principal se ejecute automáticamente al importar el archivo desde otro programa.

````python
# Ejemplo resuelto

LIMITE_PRIMERA_CATEGORIA = 150


def es_primera_categoria(peso):
    return peso >= LIMITE_PRIMERA_CATEGORIA


def main():
    peso = 170

    if es_primera_categoria(peso):
        print("Primera categoría")
    else:
        print("Otra categoría")


if __name__ == "__main__":
    main()
````

## `docstring`

Una `docstring` describe:

- El propósito de la función.
- Sus parámetros.
- El valor devuelto.
- Posibles errores o restricciones.

````python
def calcular_porcentaje(parte, total):
    """
    Calcula qué porcentaje representa una cantidad sobre un total.

    Parámetros:
        parte (int | float): cantidad parcial.
        total (int | float): cantidad total, que debe ser mayor que cero.

    Retorna:
        float: porcentaje calculado.

    Lanza:
        ValueError: si el total es igual o inferior a cero.
    """
    if total <= 0:
        raise ValueError("El total debe ser mayor que cero")

    return parte / total * 100
````

# 8. Validación de datos y gestión de errores

Un programa no debe asumir que todos los datos serán correctos.

## Errores frecuentes

- Texto cuando se espera un número.
- Valores negativos.
- Porcentajes fuera del intervalo 0–100.
- Campos vacíos.
- División entre cero.
- Datos desconocidos.
- Tipos de datos inesperados.

```python
if peso <= 0:
    return None
```

### Gestión de excepciones

```python
try:
    peso = float(input("Peso: "))
except ValueError:
    print("Debes introducir un número")
```

````python
# Ejemplo resuelto

def convertir_a_numero(texto):
    try:
        return float(texto)
    except (TypeError, ValueError):
        return None


print(convertir_a_numero("175.5"))
print(convertir_a_numero("manzana"))
````
