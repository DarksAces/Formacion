# 6. Diccionarios

Un diccionario almacena información mediante pares **clave–valor**.

```python
producto = {
    "nombre": "Pieza A",
    "unidades": 100,
    "precio": 3.5
}
```

Ventajas:

- Las claves explican el significado de los datos.
- Facilitan representar registros.
- Se pueden modificar.
- Son una base natural para trabajar después con JSON.

## Operaciones habituales

- Acceso: `diccionario["clave"]`
- Acceso seguro: `diccionario.get("clave")`
- Modificación: `diccionario["clave"] = valor`
- Recorrer claves y valores: `items()`

````python
# Ejemplo resuelto

producto = {
    "nombre": "Pieza A",
    "unidades": 100,
    "precio_unitario": 3.5
}

print(producto["nombre"])
print(producto.get("proveedor"))
print(producto.get("proveedor", "No informado"))

producto["unidades"] = 120
producto["activo"] = True

print(producto)
````

# 11. Comprensiones de listas

Una comprensión permite construir una lista de forma compacta.

## Forma básica

```python
resultado = [expresion for elemento in coleccion]
```

## Con filtro

```python
resultado = [
    expresion
    for elemento in coleccion
    if condicion
]
```

Deben utilizarse cuando mejoran la claridad. Si la lógica es compleja, un bucle normal suele ser más legible.

````python
# Ejemplos resueltos

numeros = [1, 2, 3, 4, 5]

cuadrados = [numero ** 2 for numero in numeros]
pares = [numero for numero in numeros if numero % 2 == 0]

print(cuadrados)
print(pares)
````
